# SAM 3: litter in phone frames

Test text prompts on one egocentric image at a time. Keep failures and uncertain masks for review. Run from the `ml/experiments/sam3` directory after `uv sync --extra notebook` and `uv run sam3-download`.


```python
from pathlib import Path
from PIL import Image
from IPython.display import display
from sam3_lab.inference import choose_device, overlay, segment

print("Device:", choose_device())
```

Set `image_path` to a local phone frame. Put large captures under ignored `ml/data/` and record their provenance separately.


```python
image_path = Path("../../data/example-frame.jpg")  # change to your frame
prompt = "plastic bottle"
threshold = 0.5

if not image_path.exists():
    print(f"Set image_path to an existing frame: {image_path.resolve()}")
else:
    image = Image.open(image_path).convert("RGB")
    display(image)
```

```python
if image_path.exists():
    prediction = segment(image, prompt, threshold)
    print(f"{len(prediction.scores)} instances, {prediction.elapsed_seconds:.2f}s on {prediction.device}")
    for score, box in zip(prediction.scores, prediction.boxes):
        print(f"score={score:.3f} box_xyxy={box.round(1).tolist()}")
    display(overlay(image, prediction.masks))
```
